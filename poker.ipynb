### State
- Current player
    - total number of players $m$
    - current round $r_i$
	- hand = 2 cards s.t $c_1, c_2 \in [1, 13]$
  	- pot = total bets s.t. $\sum_{i=1}^{m} b_i$
  	- earnings = $e_i$
  	- current bet = $b_i \in [0, e_i]$
	- observes
    	- cards dealt by dealer
    	- other players bets
    	- other players earnings
    	- other players experience level

### Actions
- Check
- Raise
- Fold

### Rewards
- Current pot
